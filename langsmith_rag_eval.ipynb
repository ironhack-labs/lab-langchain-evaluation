Models used throughout. **GPT-6 Luna** (`gpt-6-luna`) is OpenAI's fast, low-cost GPT-6 tier; it replaces the tutorial's `gpt-5.5` for both answering and judging. Use `gpt-5.6-luna` if you want the previous generation.

`GEN_TEMPERATURE` / `JUDGE_TEMPERATURE` are `None` (model default) because reasoning models often accept only the default temperature. The tutorial used 1 and 0; set those values if Luna accepts them.

````python
from langchain_openai import ChatOpenAI

LUNA_MODEL = "gpt-6-luna"
EMBED_MODEL = "text-embedding-3-small"
GEN_TEMPERATURE = None
JUDGE_TEMPERATURE = None

def luna(temperature=None) -> ChatOpenAI:
    kwargs = {} if temperature is None else {"temperature": temperature}
    return ChatOpenAI(model=LUNA_MODEL, **kwargs)

# Connectivity check
print(luna().invoke("Reply with the single word: ready").content)
````

## 2. Build the RAG application
### Index the documents
Load three blog posts, split them into ~250-token chunks and store them in an in-memory vector store.

````python
from langchain_community.document_loaders import WebBaseLoader
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_openai import OpenAIEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter

urls = [
    "https://lilianweng.github.io/posts/2023-06-23-agent/",
    "https://lilianweng.github.io/posts/2023-03-15-prompt-engineering/",
    "https://lilianweng.github.io/posts/2023-10-25-adv-attack-llm/",
]

docs = [WebBaseLoader(url).load() for url in urls]
docs_list = [doc for sub in docs for doc in sub]

text_splitter = RecursiveCharacterTextSplitter.from_tiktoken_encoder(
    chunk_size=250, chunk_overlap=0
)
doc_splits = text_splitter.split_documents(docs_list)

vectorstore = InMemoryVectorStore.from_documents(
    documents=doc_splits,
    embedding=OpenAIEmbeddings(model=EMBED_MODEL),
)
retriever = vectorstore.as_retriever(search_kwargs={"k": 6})
print(f"Indexed {len(doc_splits)} chunks from {len(docs_list)} pages")
````

### Generate answers
`@traceable` sends each call to LangSmith. The bot returns both the answer and the retrieved documents, because the groundedness and retrieval-relevance evaluators need the documents.

````python
from langsmith import traceable

llm = luna(GEN_TEMPERATURE)

@traceable()
def rag_bot(question: str) -> dict:
    docs = retriever.invoke(question)
    docs_string = "\n\n".join(doc.page_content for doc in docs)
    instructions = (
        "You are a helpful assistant who is good at analyzing source information "
        "and answering questions. Use the following source documents to answer "
        "the user's question. If you don't know the answer, just say that you don't know. "
        "Use three sentences maximum and keep the answer concise.\n\n"
        f"Documents:\n{docs_string}"
    )
    ai_msg = llm.invoke([
        {"role": "system", "content": instructions},
        {"role": "user", "content": question},
    ])
    return {"answer": ai_msg.content, "documents": docs}
````

````python
# Quick smoke test
out = rag_bot("What is task decomposition in LLM agents?")
print(out["answer"])
````

## 4. Define evaluators
Each evaluator is an LLM-as-judge that returns structured output: an explanation (reasoning first, which tends to improve grading) and a boolean score.

| Evaluator | Compares | Needs reference? |
|---|---|---|
| Correctness | answer ↔ reference answer | Yes |
| Relevance | answer ↔ question | No |
| Groundedness | answer ↔ retrieved docs | No |
| Retrieval relevance | retrieved docs ↔ question | No |

All four judges run on GPT-6 Luna. Each evaluator returns its score **plus the judge's explanation as a comment**, so you can read why a judge passed or failed an answer directly next to the score in LangSmith.

````python
from typing_extensions import Annotated, TypedDict

def make_grader(schema):
    return luna(JUDGE_TEMPERATURE).with_structured_output(
        schema, method="json_schema", strict=True
    )
````

### Correctness: is the answer right compared to the reference?

````python
class CorrectnessGrade(TypedDict):
    explanation: Annotated[str, ..., "Explain your reasoning for the score"]
    correct: Annotated[bool, ..., "True if the answer is correct, False otherwise."]

correctness_instructions = """You are a teacher grading a quiz. You will be given a QUESTION, the GROUND TRUTH (correct) ANSWER, and the STUDENT ANSWER.

Grading criteria:
(1) Grade the student answer based ONLY on its factual accuracy relative to the ground truth answer.
(2) Make sure the student answer does not contain any conflicting statements.
(3) It is OK if the student answer contains more information than the ground truth answer, as long as it is factually accurate relative to it.

Correctness: True means the student answer meets all the criteria. False means it does not.
Explain your reasoning step by step, then give the score. Avoid simply stating the correct answer at the outset."""

correctness_llm = make_grader(CorrectnessGrade)

def correctness(inputs: dict, outputs: dict, reference_outputs: dict) -> dict:
    """An evaluator for RAG answer accuracy."""
    content = (
        f"QUESTION: {inputs['question']}\n"
        f"GROUND TRUTH ANSWER: {reference_outputs['answer']}\n"
        f"STUDENT ANSWER: {outputs['answer']}"
    )
    grade = correctness_llm.invoke([
        {"role": "system", "content": correctness_instructions},
        {"role": "user", "content": content},
    ])
    return {"key": "correctness", "score": grade["correct"], "comment": grade["explanation"]}
````

### Relevance: does the answer address the question?

````python
class RelevanceGrade(TypedDict):
    explanation: Annotated[str, ..., "Explain your reasoning for the score"]
    relevant: Annotated[bool, ..., "Provide the score on whether the answer addresses the question"]

relevance_instructions = """You are a teacher grading a quiz. You will be given a QUESTION and a STUDENT ANSWER.

Grading criteria:
(1) The student answer is concise and relevant to the question.
(2) The student answer helps to answer the question.

Relevance: True means the student answer meets all the criteria. False means it does not.
Explain your reasoning step by step, then give the score."""

relevance_llm = make_grader(RelevanceGrade)

def relevance(inputs: dict, outputs: dict) -> dict:
    """A simple evaluator for RAG answer helpfulness."""
    content = f"QUESTION: {inputs['question']}\nSTUDENT ANSWER: {outputs['answer']}"
    grade = relevance_llm.invoke([
        {"role": "system", "content": relevance_instructions},
        {"role": "user", "content": content},
    ])
    return {"key": "relevance", "score": grade["relevant"], "comment": grade["explanation"]}
````

### Groundedness: is the answer supported by the retrieved docs (no hallucinations)?

````python
class GroundedGrade(TypedDict):
    explanation: Annotated[str, ..., "Explain your reasoning for the score"]
    grounded: Annotated[bool, ..., "Provide the score on if the answer hallucinates from the documents"]

grounded_instructions = """You are a teacher grading a quiz. You will be given FACTS and a STUDENT ANSWER.

Grading criteria:
(1) The student answer is grounded in the FACTS.
(2) The student answer does not contain "hallucinated" information outside the scope of the FACTS.

Grounded: True means the student answer meets all the criteria. False means it does not.
Explain your reasoning step by step, then give the score."""

grounded_llm = make_grader(GroundedGrade)

def groundedness(inputs: dict, outputs: dict) -> dict:
    """A simple evaluator for RAG answer groundedness."""
    doc_string = "\n\n".join(doc.page_content for doc in outputs["documents"])
    content = f"FACTS: {doc_string}\nSTUDENT ANSWER: {outputs['answer']}"
    grade = grounded_llm.invoke([
        {"role": "system", "content": grounded_instructions},
        {"role": "user", "content": content},
    ])
    return {"key": "groundedness", "score": grade["grounded"], "comment": grade["explanation"]}
````

### Retrieval relevance: are the retrieved docs relevant to the question?

````python
class RetrievalRelevanceGrade(TypedDict):
    explanation: Annotated[str, ..., "Explain your reasoning for the score"]
    relevant: Annotated[bool, ..., "True if the retrieved documents are relevant to the question, False otherwise"]

retrieval_relevance_instructions = """You are a teacher grading a quiz. You will be given a QUESTION and a set of FACTS provided by the student.

Grading criteria:
(1) Identify FACTS that are completely unrelated to the QUESTION.
(2) If the facts contain ANY keywords or semantic meaning related to the question, consider them relevant.
(3) It is OK if the facts have SOME information unrelated to the question, as long as (2) is met.

Relevance: True means the FACTS contain ANY keywords or semantic meaning related to the QUESTION. False means they are completely unrelated.
Explain your reasoning step by step, then give the score."""

retrieval_relevance_llm = make_grader(RetrievalRelevanceGrade)

def retrieval_relevance(inputs: dict, outputs: dict) -> dict:
    """An evaluator for document relevance."""
    doc_string = "\n\n".join(doc.page_content for doc in outputs["documents"])
    content = f"FACTS: {doc_string}\nQUESTION: {inputs['question']}"
    grade = retrieval_relevance_llm.invoke([
        {"role": "system", "content": retrieval_relevance_instructions},
        {"role": "user", "content": content},
    ])
    return {"key": "retrieval_relevance", "score": grade["relevant"], "comment": grade["explanation"]}
````

## 5. Run the evaluation
The target function maps a dataset example's `inputs` to the app's outputs. LangSmith runs it on every example and applies all four evaluators.

````python
def target(inputs: dict) -> dict:
    return rag_bot(inputs["question"])

experiment_results = client.evaluate(
    target,
    data=dataset_name,
    evaluators=[correctness, groundedness, relevance, retrieval_relevance],
    experiment_prefix="rag-doc-relevance-luna",
    metadata={"model": LUNA_MODEL},
)
````

````python
# Results as a DataFrame (needs pandas); the full experiment is also linked above in LangSmith
df = experiment_results.to_pandas()
feedback_cols = [c for c in df.columns if c.startswith("feedback.")]
display(df[["inputs.question", "outputs.answer", *feedback_cols]])
df[feedback_cols].mean()
````

## 7. Compare `chunk_size`, `k` and the prompt
Each configuration below rebuilds the retriever, runs the full dataset (now 15 examples) and becomes its own experiment in LangSmith, with the settings stored as metadata. The baseline is re-run too, so every experiment uses the same 15 examples and is directly comparable.

| Config | chunk_size | k | Prompt | What it tests |
|---|---|---|---|---|
| baseline | 250 | 6 | default | the tutorial's settings |
| small-k | 250 | 2 | default | less context: does retrieval miss things? |
| big-chunks | 500 | 4 | default | fewer, larger chunks (about the same amount of text) |
| strict-prompt | 250 | 6 | strict | answer only from the docs, use their wording for lists |

Each config makes about 75 Luna calls (15 answers × 5) and takes a few minutes.

````python
import pandas as pd

DEFAULT_INSTRUCTIONS = (
    "You are a helpful assistant who is good at analyzing source information "
    "and answering questions. Use the following source documents to answer "
    "the user's question. If you don't know the answer, just say that you don't know. "
    "Use three sentences maximum and keep the answer concise."
)
STRICT_INSTRUCTIONS = (
    "You are a helpful assistant who answers questions strictly from the source documents below. "
    "Only state facts that appear in the documents; if they don't contain the answer, say that you don't know. "
    "When the question asks for a list, types or steps, give the complete list using the names the documents use. "
    "Use three sentences maximum and keep the answer concise."
)
PROMPTS = {"default": DEFAULT_INSTRUCTIONS, "strict": STRICT_INSTRUCTIONS}

_stores = {}  # chunk_size -> vector store (so each chunk size is only embedded once)

def build_retriever(chunk_size: int, k: int):
    if chunk_size not in _stores:
        splitter = RecursiveCharacterTextSplitter.from_tiktoken_encoder(chunk_size=chunk_size, chunk_overlap=0)
        _stores[chunk_size] = InMemoryVectorStore.from_documents(
            documents=splitter.split_documents(docs_list),
            embedding=OpenAIEmbeddings(model=EMBED_MODEL),
        )
    return _stores[chunk_size].as_retriever(search_kwargs={"k": k})

def make_target(retriever, instructions: str):
    @traceable(name="rag_bot")
    def bot(question: str) -> dict:
        docs = retriever.invoke(question)
        docs_string = "\n\n".join(doc.page_content for doc in docs)
        ai_msg = llm.invoke([
            {"role": "system", "content": f"{instructions}\n\nDocuments:\n{docs_string}"},
            {"role": "user", "content": question},
        ])
        return {"answer": ai_msg.content, "documents": docs}
    return lambda inputs: bot(inputs["question"])

CONFIGS = [
    {"name": "baseline",      "chunk_size": 250, "k": 6, "prompt": "default"},
    {"name": "small-k",       "chunk_size": 250, "k": 2, "prompt": "default"},
    {"name": "big-chunks",    "chunk_size": 500, "k": 4, "prompt": "default"},
    {"name": "strict-prompt", "chunk_size": 250, "k": 6, "prompt": "strict"},
]

all_results, rows = {}, []
for cfg in CONFIGS:
    print(f"\n=== {cfg['name']} ===")
    results = client.evaluate(
        make_target(build_retriever(cfg["chunk_size"], cfg["k"]), PROMPTS[cfg["prompt"]]),
        data=dataset_name,
        evaluators=[correctness, groundedness, relevance, retrieval_relevance],
        experiment_prefix=f"rag-{cfg['name']}",
        metadata={"model": LUNA_MODEL, **cfg},
        max_concurrency=4,
    )
    all_results[cfg["name"]] = results
    df_cfg = results.to_pandas()
    scores = df_cfg[[c for c in df_cfg.columns if c.startswith("feedback.")]].astype(float).mean()
    rows.append({"config": cfg["name"], "experiment": results.experiment_name, **scores.round(2).to_dict()})

comparison = pd.DataFrame(rows).set_index("config")
comparison.columns = [c.replace("feedback.", "") for c in comparison.columns]
comparison
````

**Compare them side by side in LangSmith:** open *Datasets & Experiments → Lilian Weng Blogs Q&A*, tick the four `rag-…` experiments and click **Compare**. Each row is one question, so you can see exactly which question a config gets right or wrong, and click any score to read the judge's comment.

## 8. Read the judges' explanations
Every failed score, with the judge's reason, for each configuration. These are the cases worth checking by hand: sometimes the answer is wrong, sometimes the judge is.

````python
for name, results in all_results.items():
    print(f"\n=== {name} ===")
    for r in results:
        question = r["example"].inputs["question"]
        for ev in r["evaluation_results"]["results"]:
            if ev.score is not None and not ev.score:
                print(f"✗ {ev.key:<20} | {question}\n    judge: {ev.comment}\n")
````
